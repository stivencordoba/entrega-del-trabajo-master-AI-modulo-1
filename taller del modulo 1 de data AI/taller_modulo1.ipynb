### 1.1 — Listas y operaciones básicas

Tienes la siguiente lista con la esperanza de vida de 10 países (en años).

````python
# Lista dada — NO modificar
esperanzas_de_vida = [72.3, 81.5, 60.1, 78.9, 55.4, 83.2, 67.8, 74.0, 49.3, 79.6]
````

**Ejercicio 1.1** — Usando Python puro (sin Pandas ni funciones built-in como `sum()` o `max()`):
- Calcula el **promedio** de la lista usando un bucle `for`
- Encuentra el **valor máximo** usando un bucle `for`
- Encuentra el **valor mínimo** usando un bucle `for`

````python
# Tu código aquí

# utilizamos un bucle (for) para recorer la lista
# mientras realizamos una sumatoria de todos sus valores y a su vez
# evaluamos si el numero que esta en la posicion actual de la lista
# es mayor que el aterior, en caso de serlo remplazamos la variable
# (max) por dicho valor

sumatoria_de_valores = 0
max = 0
for x in range(len(esperanzas_de_vida)):
    sumatoria_de_valores += esperanzas_de_vida[x]
    if max < esperanzas_de_vida[x]:
        max = esperanzas_de_vida[x]
    

"""la sumatoria de la lista la dividimos por la cantidad de valores en la lista"""

promedio = sumatoria_de_valores/len(esperanzas_de_vida)

 
"""utilizamos el promedio para inicializar la variable (min)
que es la encargada de almacenar el numero de menor valor en
cada una de las etapas del ciclo y un condicional para verificar
si es valor actual de (min) es menor que la pocision actual de la lista, 
en caso de no ser asi modificarla"""

min = promedio
for x in range(len(esperanzas_de_vida)):
    
    if min > esperanzas_de_vida[x]:
        min = esperanzas_de_vida[x]



# Promedio
print("el promedio de los valores de la lista es:",promedio)

# Máximo
print("el numero de la lista con mayor valor es:",max)

# Mínimo
print("el numero de la lista con menor valor es:",min)
````

### 1.2 — Diccionarios

Los diccionarios son fundamentales en Data Science para representar registros.

**Ejercicio 1.2** — Crea un diccionario llamado `pais_info` que represente a Colombia con los siguientes campos:
- `nombre`: "Colombia"
- `region`: "Americas"  
- `esperanza_vida`: 76.5
- `gasto_salud_usd`: 560.3
- `nivel_ingresos`: "High"

Luego:
1. Imprime solo las **claves** del diccionario
2. Imprime solo los **valores**
3. Agrega un nuevo campo: `vacunacion_pct` con valor 82.1
4. Imprime el diccionario completo

````python
# Tu código aquí
"""vamos a crear el diccionario con los datos inicialmente dados"""

pais_info = {"nombre":"colombia","region":"america","esperanza_vida":76.5,
             "gasto_salud_usd":560.3,"nivel_ingresos":"high"}

"""vamos a utilizar los metodos (keys) y (values) para crear una lista 
con las claves y los valores respectivamente """

mis_claves = list(pais_info.keys())
mis_valores = list(pais_info.values())

#realizamos un bucle que recora la lista de claves y las imprima

print("------------- CLAVES-----------")
for z in range(len(mis_claves)):
    print("clave #",(z+1),":",mis_claves[z])

#realizamos un bucle que recora la lista de valores y las imprima

print("------------VALORES-----------")

for y in range(len(mis_valores)):
    print("valor #",(y+1),":",mis_valores[y])

#añadimos un nuevo campo al dicionario

pais_info["vacunacion_pct"] = 82.1

#imprimimos el dicionario completo atrabes de un bucle con los datos actulizados del dicionario
print("------------diccionario completo-----------")

mis_claves = list(pais_info.keys())
mis_valores = list(pais_info.values())

for y in range(len(pais_info)):
    print(mis_claves[y],":",mis_valores[y])

pais_info["vacunacion_pct"] = 82.1
````

**Ejercicio 1.3** — Crea una función llamada `clasificar_esperanza_vida` que reciba un número (años) y retorne:
- `"Alta"` si es >= 75
- `"Media"` si es entre 60 y 74.9
- `"Baja"` si es < 60

Prueba la función con al menos 5 valores distintos e imprime los resultados.

````python
# Tu código aquí

def clasificar_esperanza_vida(años):
    if años >= 75:
        return("Alta")
    elif años >= 60 and años <= 74.9:
        return("Media")
    else:
        return("Baja")
    
print(clasificar_esperanza_vida(años=1))
print(clasificar_esperanza_vida(años=80))
print(clasificar_esperanza_vida(años=65))
print(clasificar_esperanza_vida(años=44))
print(clasificar_esperanza_vida(años=100))
print(clasificar_esperanza_vida(años=77))

````

### 2.2 — Limpieza de datos

🔧 **Ejercicio 2.2** — El dataset tiene algunos valores nulos (¡datos reales siempre tienen este problema!).  
1. Rellena los valores nulos de columnas **numéricas** con la **media** de cada columna
2. Verifica que ya no queden nulos
3. Explica en una celda de texto: ¿Por qué usamos la media y no simplemente eliminamos esas filas?

````python
# Tu código aquí


# Calcular la media de cada columna numérica
print("\nPromedios de cada columna:")
for columna in df.columns:
        #Solo numérica
    if df[columna].dtype in ['float64', 'int64']:  
        media = df[columna].mean()
        # Rellenar nulos de esa columna con su media
        df[columna] = df[columna].fillna(media)

#Ver si hay nulos
print("\nNulos después:")
print(df.isnull().sum())




````

**Responde aquí** (doble clic para editar):

*¿Por qué usamos la media para rellenar nulos en vez de eliminar las filas?*

> Tu respuesta:
Eliminar filas reduce mucho el dataset y puede modificar radicalmente el análisis, remplazar con el promedio nos permite mantener un equilibrio en la informacion para obtener un resultado mas realista.
